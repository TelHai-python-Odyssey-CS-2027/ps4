````python
# Self-check - same cases as the autograder (run after your function is defined)
import copy
import math

class OneOf:
    """Several answers are acceptable."""
    def __init__(self, *options):
        self.options = options
    def __repr__(self):
        return "one of " + " / ".join(repr(o) for o in self.options)


def _same(got, expected, tol=None):
    if isinstance(expected, OneOf):
        return any(_same(got, o, tol) for o in expected.options)
    if tol is None:
        return got == expected
    if isinstance(expected, tuple):
        return (isinstance(got, tuple) and len(got) == len(expected)
                and all(_same(g, e, tol) for g, e in zip(got, expected)))
    try:
        return abs(got - expected) <= tol
    except TypeError:
        return False


def run_checks(func, func_name, groups):
    """Run every case group; print PASS/FAIL per group like the autograder."""
    earned = total = 0
    for title, points, cases in groups:
        problems = []
        for args, expected, tol, hint in cases:
            call = f"{func_name}({', '.join(repr(a) for a in args)})"
            try:
                got = func(*copy.deepcopy(args))
            except Exception as e:
                problems.append(f"    Case failed: {call}\n    Raised {type(e).__name__}: {e}")
                continue
            if not _same(got, expected, tol):
                msg = f"    Case failed: {call}\n    Expected: {expected!r}\n    Got:      {got!r}"
                if hint:
                    msg += f"\n    Hint:     {hint}"
                problems.append(msg)
        total += points
        if problems:
            print(f"[FAIL] {title}  (0/{points} pt)")
            print("\n".join(problems))
        else:
            earned += points
            print(f"[PASS] {title}  ({points}/{points} pt)")
    print(f"\nSelf-check score: {earned}/{total}")
    return earned == total


run_checks(split_at_first_digit, "split_at_first_digit", [
    ('split_at_first_digit: short symbol + one digit', 1, [
        (('H2',), ('H', 2), None, "'H2' -> ('H', 2): prefix 'H', number as an int"),
        (('O2',), ('O', 2), None, 'return the number as int, not str'),
    ]),
    ('split_at_first_digit: multi-digit number', 1, [
        (('He23',), ('He', 23), None, 'everything after the first digit is the number (23)'),
    ]),
    ('split_at_first_digit: no digit -> number 1', 1, [
        (('Fe',), ('Fe', 1), None, 'no digit found: return (formula, 1)'),
        (('F',), ('F', 1), None, 'single letter and no digit'),
    ]),
    ("split_at_first_digit: explicit 1 ('Fe1')", 1, [
        (('Fe1',), ('Fe', 1), None, 'digit present: split before it'),
    ]),
    ('split_at_first_digit: long number / long prefix', 1, [
        (('He10000',), ('He', 10000), None, 'large numbers'),
        (('Abcde1234',), ('Abcde', 1234), None, 'long prefix before the first digit'),
    ]),
])
````
